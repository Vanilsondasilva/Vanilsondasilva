```python
# =============================================================================
# Célula 5 (NOVA - CORRIGIDA): Análise Detalhada - Visitas Entre Angiotomografias
# =============================================================================
import pandas as pd
import numpy as np

print("📊 Iniciando Análise Detalhada (CORRIGIDA): Contagem de Visitas Entre Angiotomografias...")

# --- Passo 1: Mapear Datas de Angio por Paciente ---
try:
    df_angio_datas_map = df_angio_datas_COMPLETA.sort_values(by=['ID_PESSOA', 'DATA_ANGIOTOMO_OCORRENCIA'])
    angio_dates_map = df_angio_datas_map.groupby('ID_PESSOA')['DATA_ANGIOTOMO_OCORRENCIA'].apply(list).to_dict()
    print(f"   - Mapa de datas de Angiotomografia criado para {len(angio_dates_map)} pacientes.")
except NameError:
    print("🚨 ERRO: A tabela 'df_angio_datas_COMPLETA' não foi encontrada. Executa a Célula 3 primeiro.")
    raise

# --- Passo 2: Definir Função de Intervalo ---
def assign_angio_interval(row, angio_map):
    paciente_id = row['ID_PESSOA']
    event_date = row['DATA_ATENDIMENTO_FATO_PRO']

    # Verifica se a data do evento é válida (NaT)
    if pd.isna(event_date):
        return "Data_Evento_Invalida"

    angio_dates = angio_map.get(paciente_id)

    if not angio_dates:
        return "Sem_Angio_Registrada"

    if event_date < angio_dates[0]:
        return "0_Antes_1a_Angio"

    num_angios = len(angio_dates)
    for i in range(num_angios - 1):
        if angio_dates[i] <= event_date < angio_dates[i+1]:
            return f"{i+1}_Entre_{i+1}a_e_{i+2}a"

    return f"{num_angios}_Depois_{num_angios}a_Angio"

print("   - Função para definir intervalo criada.")

# --- Passo 3: Aplicar Função ao df_combinado ---
try:
    # --- CORREÇÃO AQUI ---
    # Adicionado errors='coerce' para transformar datas inválidas em NaT
    print("   - Convertendo coluna DATA_ATENDIMENTO_FATO_PRO para datetime (coerce errors)...")
    df_combinado['DATA_ATENDIMENTO_FATO_PRO'] = pd.to_datetime(df_combinado['DATA_ATENDIMENTO_FATO_PRO'], errors='coerce')

    # Conta quantos NaTs foram gerados
    nats_gerados = df_combinado['DATA_ATENDIMENTO_FATO_PRO'].isnull().sum()
    if nats_gerados > 0:
        print(f"     ... {nats_gerados} datas inválidas foram convertidas para NaT (Nulo).")
    # ---------------------

    print("   - Aplicando função de intervalo (pode demorar um pouco)...")
    df_combinado['Periodo_Angio_Detalhado'] = df_combinado.apply(
        lambda row: assign_angio_interval(row, angio_dates_map), axis=1
    )
    print("   - Coluna 'Periodo_Angio_Detalhado' criada.")
except NameError:
     print("🚨 ERRO: O DataFrame 'df_combinado' não foi encontrado. Executa a Célula 3 primeiro.")
     raise
except Exception as e:
     print(f"🚨 ERRO ao aplicar a função de intervalo: {e}")
     raise

# --- Passo 4: Criar TIPO_ATEND ---
print("   - Criando/Verificando coluna 'TIPO_ATEND'...")
if 'tipo_atendimento' not in globals():
     def tipo_atendimento(row):
          a = str(row['AGRUP_ASSISTENCIAL_G']).upper(); e = str(row['ESPECIALIDADE']).upper()
          if 'EMERGENCIA' in a: return 'EMERGENCIA'
          if 'ELETIVA' in a and 'CARDIO' in e: return 'CARDIO_ELETIVA'
          if 'ELETIVA' in a: return 'ELETIVA_OUTRA'
          return 'OUTRO'
for col in ['AGRUP_ASSISTENCIAL_G', 'ESPECIALIDADE']:
     if col in df_combinado.columns: df_combinado[col] = df_combinado[col].astype(str)
     else: df_combinado[col] = ''
df_combinado['TIPO_ATEND'] = df_combinado.apply(tipo_atendimento, axis=1)
print("   - Coluna 'TIPO_ATEND' pronta.")

# --- Passo 5 e 6: Filtrar, Agrupar e Pivotar ---
print("   - Calculando contagens detalhadas por grupo e período...")
df_visitas_analise = df_combinado[
    df_combinado['TIPO_ATEND'].isin(['EMERGENCIA', 'CARDIO_ELETIVA']) &
    ~df_combinado['Periodo_Angio_Detalhado'].isin(["Data_Evento_Invalida", "Sem_Angio_Registrada"])
].copy()
contagens = df_visitas_analise.groupby(
    ['GRUPO_ANGIO', 'TIPO_ATEND', 'Periodo_Angio_Detalhado']
).size().reset_index(name='Contagem')
resumo_detalhado = pd.pivot_table(
    contagens, values='Contagem', index='GRUPO_ANGIO',
    columns=['TIPO_ATEND', 'Periodo_Angio_Detalhado'], fill_value=0
)
resumo_detalhado.columns = ['_'.join(col).strip() for col in resumo_detalhado.columns.values]
resumo_detalhado = resumo_detalhado[sorted(resumo_detalhado.columns)]
print("   - Contagens calculadas e tabela pivotada.")

# --- Passo 7: Adicionar Contagem de Pacientes ---
pacientes_por_grupo = df_combinado.groupby('GRUPO_ANGIO')['ID_PESSOA'].nunique()
resumo_detalhado['Pacientes_Unicos_no_Grupo'] = pacientes_por_grupo
resumo_detalhado = resumo_detalhado[['Pacientes_Unicos_no_Grupo'] + [col for col in resumo_detalhado.columns if col != 'Pacientes_Unicos_no_Grupo']]

# --- Exibir Tabela Final ---
print("\n--- 📊 Tabela Resumo Detalhada: Visitas Entre Angiotomografias por Grupo ---")
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 500)
display(resumo_detalhado)
pd.reset_option('display.max_rows')
pd.reset_option('display.max_columns')

print("\n✅ Análise Detalhada (CORRIGIDA) concluída.")
```

```python
# =============================================================================
# Célula 6 (NOVA - CORRIGIDA v5 - Vectorizada): Análise de Internações
# =============================================================================
import pandas as pd
import numpy as np
import re

print("🏥 Iniciando Análise de Internações Entre Angiotomografias (v5 - Vectorizada)...")

# --- Pré-requisitos ---
try:
    df_combinado
    df_angio_datas_COMPLETA # Precisamos desta tabela agora
except NameError:
    print("🚨 ERRO: 'df_combinado' ou 'df_angio_datas_COMPLETA' não encontrados. Executa a Célula 3 primeiro.")
    raise

# --- Passo 1: Identificar Internações Válidas e Calcular Duração ---
print("   - Identificando internações válidas e calculando duração...")
df_internacoes = df_combinado.copy()
df_internacoes['DATA_INTERNACAO'] = pd.to_datetime(df_internacoes['DATA_INTERNACAO'], errors='coerce')
df_internacoes['DATA_ALTA'] = pd.to_datetime(df_internacoes['DATA_ALTA'], errors='coerce')
mask_internacao_valida = (df_internacoes['DATA_INTERNACAO'].notna() & df_internacoes['DATA_ALTA'].notna() & (df_internacoes['DATA_ALTA'] >= df_internacoes['DATA_INTERNACAO']))
df_internacoes = df_internacoes[mask_internacao_valida].copy()

if df_internacoes.empty:
    print("   - Nenhuma internação válida encontrada no DataFrame.")
else:
    print(f"   - {df_internacoes.shape[0]} registos de internação válidos encontrados.")
    df_internacoes['Duracao_Internacao'] = (df_internacoes['DATA_ALTA'] - df_internacoes['DATA_INTERNACAO']).dt.days + 1

    # --- Passo 3: Juntar Datas de Angio como Colunas ---
    print("   - Juntando datas das Angiotomografias (1ª, 2ª, etc.) ao DataFrame...")
    # Pivotar df_angio_datas para ter ID_PESSOA como índice e ORDEM_ANGIOTOMO como colunas
    df_angio_pivot = df_angio_datas_COMPLETA.pivot(
        index='ID_PESSOA',
        columns='ORDEM_ANGIOTOMO',
        values='DATA_ANGIOTOMO_OCORRENCIA'
    )
    # Renomear colunas para Data_Angio_1, Data_Angio_2, etc.
    df_angio_pivot.columns = [f'Data_Angio_{col}' for col in df_angio_pivot.columns]

    # Juntar estas datas ao df_internacoes
    df_internacoes = pd.merge(df_internacoes, df_angio_pivot, on='ID_PESSOA', how='left')
    print("   - Datas das Angios adicionadas.")

    # --- Passo 4: Atribuir Período (Vectorizado com np.select) ---
    print("   - Atribuindo período da Angiotomografia (vectorizado)...")

    # Definir as condições e escolhas para np.select
    # Começamos do último intervalo para o primeiro
    conditions = []
    choices = []

    # Máximo de Angios a considerar (pode ajustar se necessário)
    max_angios = 5 # Baseado nos teus grupos

    # Condição: Depois da Última Angio (ex: Depois da 5ª)
    if f'Data_Angio_{max_angios}' in df_internacoes.columns:
         conditions.append(df_internacoes['DATA_INTERNACAO'] >= df_internacoes[f'Data_Angio_{max_angios}'])
         choices.append(f"{max_angios}_Depois_{max_angios}a_Angio")

    # Condições: Entre Angios (ex: Entre 4ª e 5ª, ..., Entre 1ª e 2ª)
    for i in range(max_angios - 1, 0, -1):
         col_angio_i = f'Data_Angio_{i}'
         col_angio_i_mais_1 = f'Data_Angio_{i+1}'
         if col_angio_i in df_internacoes.columns and col_angio_i_mais_1 in df_internacoes.columns:
              conditions.append(
                   (df_internacoes['DATA_INTERNACAO'] >= df_internacoes[col_angio_i]) &
                   (df_internacoes['DATA_INTERNACAO'] < df_internacoes[col_angio_i_mais_1]) &
                   df_internacoes[col_angio_i_mais_1].notna() # Garante que a Angio seguinte existe
              )
              choices.append(f"{i}_Entre_{i}a_e_{i+1}a")

    # Condição: Antes da 1ª Angio
    if 'Data_Angio_1' in df_internacoes.columns:
         conditions.append(df_internacoes['DATA_INTERNACAO'] < df_internacoes['Data_Angio_1'])
         choices.append("0_Antes_1a_Angio")

    # Aplicar np.select
    df_internacoes['Periodo_Angio_Detalhado'] = np.select(conditions, choices, default='Sem_Angio_Registrada_ou_Erro') # Default caso não caia em nenhuma

    # Limpeza (remover casos onde não encontramos angio ou houve erro)
    df_internacoes = df_internacoes[df_internacoes['Periodo_Angio_Detalhado'] != 'Sem_Angio_Registrada_ou_Erro']
    print("   - Períodos atribuídos.")

    # --- Passo 5: Calcular Dias Desde Início do Período (Vectorizado) ---
    print("   - Calculando dias desde o início do período (vectorizado)...")

    conditions_dias = []
    choices_dias = []

    # Iterar pelos possíveis inícios de período (Angio 1 a max_angios)
    for i in range(1, max_angios + 1):
        col_angio_i = f'Data_Angio_{i}'
        if col_angio_i in df_internacoes.columns:
            # Condição: O período começa *após* ou *na* Angio i
            # (Ex: '1_Entre_1a_e_2a', '2_Entre_2a_e_3a', '5_Depois_5a_Angio')
            conditions_dias.append(df_internacoes['Periodo_Angio_Detalhado'].str.match(f"^{i}_"))
            # Escolha: Diferença em dias
            choices_dias.append((df_internacoes['DATA_INTERNACAO'] - df_internacoes[col_angio_i]).dt.days)

    df_internacoes['Dias_Desde_Inicio_Periodo'] = np.select(conditions_dias, choices_dias, default=np.nan) # Default é NaN (para '0_Antes_1a_Angio')
    print("   - Dias desde o início calculados.")

    # --- Passo 6: Agrupar e Calcular Médias ---
    print("   - Agrupando e calculando métricas...")
    resumo_internacoes = df_internacoes.groupby(['GRUPO_ANGIO', 'Periodo_Angio_Detalhado']).agg(
        Qtd_Internacoes=('ID_PESSOA', 'size'),
        Media_Dias_Internado=('Duracao_Internacao', 'mean'),
        Media_Dias_Desde_Inicio_Periodo=('Dias_Desde_Inicio_Periodo', 'mean')
    ).reset_index()

    # --- Passo 7: Formatar Tabela ---
    pacientes_por_grupo = df_combinado.groupby('GRUPO_ANGIO')['ID_PESSOA'].nunique().reset_index()
    pacientes_por_grupo = pacientes_por_grupo.rename(columns={'ID_PESSOA':'Pacientes_Unicos_no_Grupo'})
    resumo_internacoes_final = pd.merge(resumo_internacoes, pacientes_por_grupo, on='GRUPO_ANGIO', how='left')
    resumo_internacoes_final = resumo_internacoes_final[[
        'GRUPO_ANGIO', 'Pacientes_Unicos_no_Grupo', 'Periodo_Angio_Detalhado',
        'Qtd_Internacoes', 'Media_Dias_Internado', 'Media_Dias_Desde_Inicio_Periodo'
    ]].sort_values(by=['GRUPO_ANGIO', 'Periodo_Angio_Detalhado'])
    print("   - Métricas calculadas.")

    # --- Exibir Tabela Final ---
    print("\n--- 🏥 Tabela Resumo: Internações Entre Angiotomografias por Grupo ---")
    pd.set_option('display.max_rows', 500); pd.set_option('display.max_columns', 500)
    display(resumo_internacoes_final.round(1))
    pd.reset_option('display.max_rows'); pd.reset_option('display.max_columns')

print("\n✅ Análise de Internações (v5 - Vectorizada) concluída.")
```

```python
# =============================================================================
# BLOCO DE ANÁLISE COMBINADO 1: INFARTO
# (Execute este bloco inteiro - Opera sobre 'df' = df_combinado)
# =============================================================================
import pandas as pd
import numpy as np
import plotly.express as px

print("==================================================================================")
print("🚀 INICIANDO PIPELINE (COMBINADO 1/5) PARA: INFARTO")
print("==================================================================================")

# -----------------------------------------------------------------------------
# Célula 4 (Novo Pipeline): Filtrar Pacientes e Definir Tipos
# -----------------------------------------------------------------------------
print("\n🔍 Iniciando Célula 4 (Infarto - Combinado): Filtrando pacientes...")

keywords_infarto = ['INFARTO AGUDO']
pattern_infarto = '|'.join(keywords_infarto)
mask_infarto_any = df['DESCRICAO_CID'].str.contains(pattern_infarto, case=False, na=False)
ids_infarto = df.loc[mask_infarto_any, 'ID_PESSOA'].unique()
print(f"🫀 IDs únicos com registro de INFARTO (em todo df): {len(ids_infarto)}")

df_jornada_infarto = df[df['ID_PESSOA'].isin(ids_infarto)].copy()
print(f"📂 Registros desses pacientes (toda jornada): {df_jornada_infarto.shape[0]:,}")

for c in ['AGRUP_ASSISTENCIAL_G', 'ESPECIALIDADE', 'DESCRICAO_CID', 'SERVICO']:
    if c in df_jornada_infarto.columns: df_jornada_infarto[c] = df_jornada_infarto[c].astype(str)
    else: df_jornada_infarto[c] = '' # Adiciona coluna vazia se faltar

def tipo_atendimento(row):
    a = str(row['AGRUP_ASSISTENCIAL_G']).upper(); e = str(row['ESPECIALIDADE']).upper()
    if 'EMERGENCIA' in a: return 'EMERGENCIA'
    if 'ELETIVA' in a and 'CARDIO' in e: return 'CARDIO_ELETIVA'
    if 'ELETIVA' in a: return 'ELETIVA_OUTRA'
    return 'OUTRO'

df_jornada_infarto['TIPO_ATEND'] = df_jornada_infarto.apply(tipo_atendimento, axis=1)
print(" ... Coluna 'TIPO_ATEND' criada.")
print(f"✅ 'df_jornada_infarto' está pronto.")

# -----------------------------------------------------------------------------
# Célula 5: Busca por procedimentos-chave (Angio/Cateter/Stent)
# -----------------------------------------------------------------------------
print("\nIniciando Célula 5 (Infarto - Combinado): Busca por procedimentos-chave...")

keywords_procedimentos = ['ANGIOTOMOGRAFIA CORONARIANA', 'CATETERISMO CARDIACO', 'IMPLANTE DE STENT CORONARIO']
pattern = '|'.join(keywords_procedimentos)

mask = df_jornada_infarto['SERVICO'].str.contains(pattern, case=False, na=False)
df_proc = df_jornada_infarto[mask].copy()
df_proc_unicos = df_proc.drop_duplicates(subset=['ID_PESSOA', 'DATA_ATENDIMENTO_FATO_PRO', 'SERVICO'])

if df_proc_unicos.empty: print(" ... Nenhum procedimento-chave encontrado.")
else: print(f" ... SUCESSO! Encontrados {df_proc_unicos.shape[0]} procedimentos-chave únicos.")
print("✅ Célula 5 concluída.")

# -----------------------------------------------------------------------------
# Célula 6: Cronologia Real (Timeline em Dias)
# -----------------------------------------------------------------------------
print("\n⏳ Iniciando Célula 6 (Infarto - Combinado): Construindo a Linha do Tempo Real...")

df_base_infarto = df_jornada_infarto.copy()
df_base_infarto['DATA_ATENDIMENTO_FATO_PRO'] = pd.to_datetime(df_base_infarto['DATA_ATENDIMENTO_FATO_PRO'], errors='coerce')
df_base_infarto = df_base_infarto.dropna(subset=['DATA_ATENDIMENTO_FATO_PRO']) # Remove linhas onde a data falhou

mask_infarto_temp = df_base_infarto['DESCRICAO_CID'].str.contains(pattern_infarto, case=False, na=False)
df_infarto_eventos = df_base_infarto[mask_infarto_temp]
event_0_dates = df_infarto_eventos.groupby('ID_PESSOA')['DATA_ATENDIMENTO_FATO_PRO'].min().reset_index()
event_0_dates.columns = ['ID_PESSOA', 'Data_Evento_0']

df_jornada_timeline_infarto = pd.merge(df_base_infarto, event_0_dates, on='ID_PESSOA', how='left')
# Apenas pacientes que tiveram o evento infarto terão Data_Evento_0 não nula
df_jornada_timeline_infarto = df_jornada_timeline_infarto.dropna(subset=['Data_Evento_0'])

df_jornada_timeline_infarto['Dias_Desde_Evento_0'] = (df_jornada_timeline_infarto['DATA_ATENDIMENTO_FATO_PRO'] - df_jornada_timeline_infarto['Data_Evento_0']).dt.days

mask_proc = df_jornada_timeline_infarto['SERVICO'].str.contains(pattern, case=False, na=False)
mask_emerg = df_jornada_timeline_infarto['TIPO_ATEND'] == 'EMERGENCIA'
mask_cardio_elet = df_jornada_timeline_infarto['TIPO_ATEND'] == 'CARDIO_ELETIVA'
mask_infarto = df_jornada_timeline_infarto['DESCRICAO_CID'].str.contains(pattern_infarto, case=False, na=False)

mask_key_events = mask_infarto | mask_proc | mask_emerg | mask_cardio_elet
df_timeline_filtrada = df_jornada_timeline_infarto[mask_key_events].copy()

conditions = [
    df_timeline_filtrada['SERVICO'].str.contains('IMPLANTE DE STENT', case=False, na=False),
    df_timeline_filtrada['SERVICO'].str.contains('CATETERISMO CARDIACO', case=False, na=False),
    df_timeline_filtrada['SERVICO'].str.contains('ANGIOTOMOGRAFIA CORONARIANA', case=False, na=False),
    (df_timeline_filtrada['DESCRICAO_CID'].str.contains(pattern_infarto, case=False, na=False)) & (df_timeline_filtrada['DATA_ATENDIMENTO_FATO_PRO'] == df_timeline_filtrada['Data_Evento_0']),
    df_timeline_filtrada['TIPO_ATEND'] == 'EMERGENCIA',
    df_timeline_filtrada['TIPO_ATEND'] == 'CARDIO_ELETIVA'
]
choices = ['PROC: Implante de Stent', 'PROC: Cateterismo Cardiaco', 'PROC: Angiotomografia COR', 'EVENTO: Registro Infarto', 'VISITA: Emergência', 'VISITA: Cardio Eletiva']
df_timeline_filtrada['Evento_Label'] = np.select(conditions, choices, default='Outro')

# Manter a coluna GRUPO_ANGIO na tabela final!
colunas_finais = ['ID_PESSOA', 'GRUPO_ANGIO', 'DATA_ATENDIMENTO_FATO_PRO', 'Dias_Desde_Evento_0', 'Evento_Label']
# NOME DISTINTO PARA GUARDAR O RESULTADO
df_timeline_final_view_infarto_COMBINED = df_timeline_filtrada[colunas_finais]
df_timeline_final_view_infarto_COMBINED = df_timeline_final_view_infarto_COMBINED[df_timeline_final_view_infarto_COMBINED['Evento_Label'] != 'Outro'].copy()
df_timeline_final_view_infarto_COMBINED = df_timeline_final_view_infarto_COMBINED.sort_values(by=['ID_PESSOA', 'Dias_Desde_Evento_0'])
df_timeline_final_view_infarto_COMBINED = df_timeline_final_view_infarto_COMBINED.drop_duplicates(subset=['ID_PESSOA', 'DATA_ATENDIMENTO_FATO_PRO', 'Evento_Label'])

print(f" ... Tabela de cronologia 'df_timeline_final_view_infarto_COMBINED' criada com {df_timeline_final_view_infarto_COMBINED.shape[0]} eventos-chave.")
print("✅ Célula 6 concluída.")

# -----------------------------------------------------------------------------
# Célula 7: Visualização da Cronologia (Timeline Plot)
# -----------------------------------------------------------------------------
print("\n📊 Iniciando Célula 7 (Infarto - Combinado): Geração do Gráfico de Cronologia...")
# Adiciona GRUPO_ANGIO ao hover e usa-o para cor/facet? (Pode ficar poluído)
df_plot = df_timeline_final_view_infarto_COMBINED.copy()
df_plot['ID_PESSOA'] = df_plot['ID_PESSOA'].astype(str) # ID como string

# Calcular range dinâmico
min_dias = df_plot['Dias_Desde_Evento_0'].min(); max_dias = df_plot['Dias_Desde_Evento_0'].max()
min_dias_com_margem = min_dias - 30; max_dias_com_margem = max_dias + 30

fig = px.scatter(
    data_frame=df_plot,
    x='Dias_Desde_Evento_0', y='ID_PESSOA', color='Evento_Label',
    hover_data=['DATA_ATENDIMENTO_FATO_PRO', 'Evento_Label', 'Dias_Desde_Evento_0', 'GRUPO_ANGIO'], # Adicionado GRUPO_ANGIO
    labels={'Dias_Desde_Evento_0': 'Dias desde o Evento Zero (1º Infarto)', 'ID_PESSOA': 'Paciente', 'Evento_Label': 'Tipo de Evento'},
    title=f'Cronologia (Infarto) - Todos os Grupos ({df_plot["ID_PESSOA"].nunique()} pacientes)'
)
fig.add_vline(x=0, line_width=2, line_dash="dash", line_color="red", annotation_text="Dia 0 (Infarto)", annotation_position="bottom right")
fig.update_yaxes(type='category', showticklabels=False, title=f'Pacientes ({df_plot["ID_PESSOA"].nunique()} no total)')
fig.update_layout(yaxis={'categoryorder':'total ascending'}, xaxis_range=[min_dias_com_margem, max_dias_com_margem], height=800)

print(" ... Gráfico gerado. (Será exibido abaixo)")
fig.show()
print("✅ Célula 7 concluída.")

# -----------------------------------------------------------------------------
# Célula 9: Cálculo de Métricas Preventivas (APENAS CÁLCULO, SEM MÉDIAS GERAIS)
# -----------------------------------------------------------------------------
print("\n🚨 Iniciando Célula 9 (Infarto - Combinado): Calculando KPIs Preventivos...")
# Usa a tabela de cronologia COMBINED
df_pre_infarto = df_timeline_final_view_infarto_COMBINED[
    df_timeline_final_view_infarto_COMBINED['Dias_Desde_Evento_0'] < 0
].copy()

if df_pre_infarto.empty:
    print("  - Nenhum evento registrado antes do 'Dia Zero' para os pacientes com Infarto.")
    # Cria df vazio para evitar erros, incluindo GRUPO_ANGIO
    df_kpi_preventivo_infarto_COMBINED = pd.DataFrame(columns=['ID_PESSOA', 'GRUPO_ANGIO', 'Qtd_Cardio_Eletiva_Pre_Infarto', 'Qtd_Emergencia_Pre_Infarto', 'Qtd_Procedimentos_Pre_Infarto'])
else:
    print(f"  - {df_pre_infarto.shape[0]} eventos pré-infarto encontrados para análise.")

    # Agrupa por ID_PESSOA E GRUPO_ANGIO para calcular KPIs
    kpi_cardio_pre = df_pre_infarto[df_pre_infarto['Evento_Label'] == 'VISITA: Cardio Eletiva'].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Cardio_Eletiva_Pre_Infarto')
    kpi_emerg_pre = df_pre_infarto[df_pre_infarto['Evento_Label'] == 'VISITA: Emergência'].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Emergencia_Pre_Infarto')
    kpi_proc_pre = df_pre_infarto[df_pre_infarto['Evento_Label'].str.contains('PROC:')].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Procedimentos_Pre_Infarto')

    # Base com todos os pacientes DESTE GRUPO DE DOENÇA e seus GRUPOS ANGIO
    df_pacientes_infarto = df_timeline_final_view_infarto_COMBINED[['ID_PESSOA', 'GRUPO_ANGIO']].drop_duplicates()

    # Faz merge mantendo GRUPO_ANGIO
    df_kpi_preventivo_infarto_COMBINED = df_pacientes_infarto.merge(kpi_cardio_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')
    df_kpi_preventivo_infarto_COMBINED = df_kpi_preventivo_infarto_COMBINED.merge(kpi_emerg_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')
    df_kpi_preventivo_infarto_COMBINED = df_kpi_preventivo_infarto_COMBINED.merge(kpi_proc_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')

    cols_kpi = ['Qtd_Cardio_Eletiva_Pre_Infarto', 'Qtd_Emergencia_Pre_Infarto', 'Qtd_Procedimentos_Pre_Infarto']
    df_kpi_preventivo_infarto_COMBINED[cols_kpi] = df_kpi_preventivo_infarto_COMBINED[cols_kpi].fillna(0).astype(int)

    print("\n--- 🚨 Métricas Preventivas por Paciente (Comportamento Pré-Infarto) ---")
    # Mostra apenas as primeiras linhas como exemplo
    display(df_kpi_preventivo_infarto_COMBINED.head())

# Não calculamos as médias GERAIS aqui. Faremos isso nas Células 10/11 modificadas.
print("✅ Célula 9 concluída. Tabela 'df_kpi_preventivo_infarto_COMBINED' criada.")

print("\n==================================================================================")
print("🏁 PIPELINE (COMBINADO 1/5) PARA INFARTO CONCLUÍDO 🏁")
print("==================================================================================")
```

```python
# =============================================================================
# BLOCO DE ANÁLISE COMBINADO 4: INSUFICIÊNCIA
# (Execute este bloco inteiro - Opera sobre 'df' = df_combinado)
# =============================================================================
import pandas as pd
import numpy as np
import plotly.express as px

print("==================================================================================")
print("🚀 INICIANDO PIPELINE (COMBINADO 4/5) PARA: INSUFICIÊNCIA")
print("==================================================================================")

# -----------------------------------------------------------------------------
# Célula 4 (Novo Pipeline): Filtrar Pacientes e Definir Tipos
# -----------------------------------------------------------------------------
print("\n🔍 Iniciando Célula 4 (Insuficiência - Combinado): Filtrando pacientes...")

keywords_insuficiencia = [
    'INSUFICIENCIA (DA VALVA) AORTICA',
    'INSUFICIENCIA CARDIACA NAO ESPECIFICADA',
    'DOENÇA CARDIACA HIPERTENSIVA SEM INSUFICIENCIA',
    'INSUFICIENCIA CARDIACA CONGESTIVA',
    'DOENÇA CARDIACA HIPERTENSIVA COM INSUFICIENCIA',
    'INSUFICIENCIA (DA VALVA) MITRAL'
]
pattern_insuficiencia = '|'.join(keywords_insuficiencia)
mask_insuficiencia_any = df['DESCRICAO_CID'].str.contains(pattern_insuficiencia, case=False, na=False)
ids_insuficiencia = df.loc[mask_insuficiencia_any, 'ID_PESSOA'].unique()
print(f"🫀 IDs únicos com registro de INSUFICIÊNCIA (em todo df): {len(ids_insuficiencia)}")

df_jornada_insuficiencia = df[df['ID_PESSOA'].isin(ids_insuficiencia)].copy()
print(f"📂 Registros desses pacientes (toda jornada): {df_jornada_insuficiencia.shape[0]:,}")

for c in ['AGRUP_ASSISTENCIAL_G', 'ESPECIALIDADE', 'DESCRICAO_CID', 'SERVICO']:
    if c in df_jornada_insuficiencia.columns: df_jornada_insuficiencia[c] = df_jornada_insuficiencia[c].astype(str)
    else: df_jornada_insuficiencia[c] = ''

# Função tipo_atendimento já deve existir na memória
df_jornada_insuficiencia['TIPO_ATEND'] = df_jornada_insuficiencia.apply(tipo_atendimento, axis=1)
print(" ... Coluna 'TIPO_ATEND' criada.")
print(f"✅ 'df_jornada_insuficiencia' está pronto.")

# -----------------------------------------------------------------------------
# Célula 5: Busca por procedimentos-chave (Angio/Cateter/Stent)
# -----------------------------------------------------------------------------
print("\nIniciando Célula 5 (Insuficiência - Combinado): Busca por procedimentos-chave...")

keywords_procedimentos = ['ANGIOTOMOGRAFIA CORONARIANA', 'CATETERISMO CARDIACO', 'IMPLANTE DE STENT CORONARIO']
pattern = '|'.join(keywords_procedimentos) # Reutilizado na Célula 6

mask = df_jornada_insuficiencia['SERVICO'].str.contains(pattern, case=False, na=False)
df_proc = df_jornada_insuficiencia[mask].copy()
df_proc_unicos = df_proc.drop_duplicates(subset=['ID_PESSOA', 'DATA_ATENDIMENTO_FATO_PRO', 'SERVICO'])

if df_proc_unicos.empty: print(" ... Nenhum procedimento-chave encontrado.")
else: print(f" ... SUCESSO! Encontrados {df_proc_unicos.shape[0]} procedimentos-chave únicos.")
print("✅ Célula 5 concluída.")

# -----------------------------------------------------------------------------
# Célula 6: Cronologia Real (Timeline em Dias)
# -----------------------------------------------------------------------------
print("\n⏳ Iniciando Célula 6 (Insuficiência - Combinado): Construindo a Linha do Tempo Real...")

df_base_insuficiencia = df_jornada_insuficiencia.copy()
df_base_insuficiencia['DATA_ATENDIMENTO_FATO_PRO'] = pd.to_datetime(df_base_insuficiencia['DATA_ATENDIMENTO_FATO_PRO'], errors='coerce')
df_base_insuficiencia = df_base_insuficiencia.dropna(subset=['DATA_ATENDIMENTO_FATO_PRO'])

mask_insuficiencia_temp = df_base_insuficiencia['DESCRICAO_CID'].str.contains(pattern_insuficiencia, case=False, na=False)
df_insuficiencia_eventos = df_base_insuficiencia[mask_insuficiencia_temp]
event_0_dates = df_insuficiencia_eventos.groupby('ID_PESSOA')['DATA_ATENDIMENTO_FATO_PRO'].min().reset_index()
event_0_dates.columns = ['ID_PESSOA', 'Data_Evento_0']

df_jornada_timeline_insuficiencia = pd.merge(df_base_insuficiencia, event_0_dates, on='ID_PESSOA', how='left')
df_jornada_timeline_insuficiencia = df_jornada_timeline_insuficiencia.dropna(subset=['Data_Evento_0'])

df_jornada_timeline_insuficiencia['Dias_Desde_Evento_0'] = (df_jornada_timeline_insuficiencia['DATA_ATENDIMENTO_FATO_PRO'] - df_jornada_timeline_insuficiencia['Data_Evento_0']).dt.days

mask_proc = df_jornada_timeline_insuficiencia['SERVICO'].str.contains(pattern, case=False, na=False)
mask_emerg = df_jornada_timeline_insuficiencia['TIPO_ATEND'] == 'EMERGENCIA'
mask_cardio_elet = df_jornada_timeline_insuficiencia['TIPO_ATEND'] == 'CARDIO_ELETIVA'
mask_insuficiencia = df_jornada_timeline_insuficiencia['DESCRICAO_CID'].str.contains(pattern_insuficiencia, case=False, na=False)

mask_key_events = mask_insuficiencia | mask_proc | mask_emerg | mask_cardio_elet
df_timeline_filtrada = df_jornada_timeline_insuficiencia[mask_key_events].copy()

conditions = [
    df_timeline_filtrada['SERVICO'].str.contains('IMPLANTE DE STENT', case=False, na=False),
    df_timeline_filtrada['SERVICO'].str.contains('CATETERISMO CARDIACO', case=False, na=False),
    df_timeline_filtrada['SERVICO'].str.contains('ANGIOTOMOGRAFIA CORONARIANA', case=False, na=False),
    (df_timeline_filtrada['DESCRICAO_CID'].str.contains(pattern_insuficiencia, case=False, na=False)) & (df_timeline_filtrada['DATA_ATENDIMENTO_FATO_PRO'] == df_timeline_filtrada['Data_Evento_0']),
    df_timeline_filtrada['TIPO_ATEND'] == 'EMERGENCIA',
    df_timeline_filtrada['TIPO_ATEND'] == 'CARDIO_ELETIVA'
]
choices = ['PROC: Implante de Stent', 'PROC: Cateterismo Cardiaco', 'PROC: Angiotomografia COR', 'EVENTO: Registro Insuficiência', 'VISITA: Emergência', 'VISITA: Cardio Eletiva']
df_timeline_filtrada['Evento_Label'] = np.select(conditions, choices, default='Outro')

colunas_finais = ['ID_PESSOA', 'GRUPO_ANGIO', 'DATA_ATENDIMENTO_FATO_PRO', 'Dias_Desde_Evento_0', 'Evento_Label']
# NOME DISTINTO PARA GUARDAR O RESULTADO
df_timeline_final_view_insuficiencia_COMBINED = df_timeline_filtrada[colunas_finais]
df_timeline_final_view_insuficiencia_COMBINED = df_timeline_final_view_insuficiencia_COMBINED[df_timeline_final_view_insuficiencia_COMBINED['Evento_Label'] != 'Outro'].copy()
df_timeline_final_view_insuficiencia_COMBINED = df_timeline_final_view_insuficiencia_COMBINED.sort_values(by=['ID_PESSOA', 'Dias_Desde_Evento_0'])
df_timeline_final_view_insuficiencia_COMBINED = df_timeline_final_view_insuficiencia_COMBINED.drop_duplicates(subset=['ID_PESSOA', 'DATA_ATENDIMENTO_FATO_PRO', 'Evento_Label'])

print(f" ... Tabela de cronologia 'df_timeline_final_view_insuficiencia_COMBINED' criada com {df_timeline_final_view_insuficiencia_COMBINED.shape[0]} eventos-chave.")
print("✅ Célula 6 concluída.")

# -----------------------------------------------------------------------------
# Célula 7: Visualização da Cronologia (Timeline Plot)
# -----------------------------------------------------------------------------
print("\n📊 Iniciando Célula 7 (Insuficiência - Combinado): Geração do Gráfico de Cronologia...")
df_plot = df_timeline_final_view_insuficiencia_COMBINED.copy()
df_plot['ID_PESSOA'] = df_plot['ID_PESSOA'].astype(str)

min_dias = df_plot['Dias_Desde_Evento_0'].min(); max_dias = df_plot['Dias_Desde_Evento_0'].max()
min_dias_com_margem = min_dias - 30; max_dias_com_margem = max_dias + 30

fig = px.scatter(
    data_frame=df_plot,
    x='Dias_Desde_Evento_0', y='ID_PESSOA', color='Evento_Label',
    hover_data=['DATA_ATENDIMENTO_FATO_PRO', 'Evento_Label', 'Dias_Desde_Evento_0', 'GRUPO_ANGIO'],
    labels={'Dias_Desde_Evento_0': 'Dias desde o Evento Zero (1ª Insuficiência)', 'ID_PESSOA': 'Paciente', 'Evento_Label': 'Tipo de Evento'},
    title=f'Cronologia (Insuficiência) - Todos os Grupos ({df_plot["ID_PESSOA"].nunique()} pacientes)'
)
fig.add_vline(x=0, line_width=2, line_dash="dash", line_color="red", annotation_text="Dia 0 (Insuficiência)", annotation_position="bottom right")
fig.update_yaxes(type='category', showticklabels=False, title=f'Pacientes ({df_plot["ID_PESSOA"].nunique()} no total)')
fig.update_layout(yaxis={'categoryorder':'total ascending'}, xaxis_range=[min_dias_com_margem, max_dias_com_margem], height=800)

print(" ... Gráfico gerado. (Será exibido abaixo)")
fig.show()
print("✅ Célula 7 concluída.")

# -----------------------------------------------------------------------------
# Célula 9: Cálculo de Métricas Preventivas (APENAS CÁLCULO, SEM MÉDIAS GERAIS)
# -----------------------------------------------------------------------------
print("\n🚨 Iniciando Célula 9 (Insuficiência - Combinado): Calculando KPIs Preventivos...")
df_pre_insuficiencia = df_timeline_final_view_insuficiencia_COMBINED[
    df_timeline_final_view_insuficiencia_COMBINED['Dias_Desde_Evento_0'] < 0
].copy()

if df_pre_insuficiencia.empty:
    print("  - Nenhum evento registrado antes do 'Dia Zero' para os pacientes com Insuficiência.")
    df_kpi_preventivo_insuficiencia_COMBINED = pd.DataFrame(columns=['ID_PESSOA', 'GRUPO_ANGIO', 'Qtd_Cardio_Eletiva_Pre_Insuficiencia', 'Qtd_Emergencia_Pre_Insuficiencia', 'Qtd_Procedimentos_Pre_Insuficiencia'])
else:
    print(f"  - {df_pre_insuficiencia.shape[0]} eventos pré-insuficiência encontrados para análise.")

    kpi_cardio_pre = df_pre_insuficiencia[df_pre_insuficiencia['Evento_Label'] == 'VISITA: Cardio Eletiva'].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Cardio_Eletiva_Pre_Insuficiencia')
    kpi_emerg_pre = df_pre_insuficiencia[df_pre_insuficiencia['Evento_Label'] == 'VISITA: Emergência'].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Emergencia_Pre_Insuficiencia')
    kpi_proc_pre = df_pre_insuficiencia[df_pre_insuficiencia['Evento_Label'].str.contains('PROC:')].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Procedimentos_Pre_Insuficiencia')

    df_pacientes_insuficiencia = df_timeline_final_view_insuficiencia_COMBINED[['ID_PESSOA', 'GRUPO_ANGIO']].drop_duplicates()

    df_kpi_preventivo_insuficiencia_COMBINED = df_pacientes_insuficiencia.merge(kpi_cardio_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')
    df_kpi_preventivo_insuficiencia_COMBINED = df_kpi_preventivo_insuficiencia_COMBINED.merge(kpi_emerg_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')
    df_kpi_preventivo_insuficiencia_COMBINED = df_kpi_preventivo_insuficiencia_COMBINED.merge(kpi_proc_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')

    cols_kpi = ['Qtd_Cardio_Eletiva_Pre_Insuficiencia', 'Qtd_Emergencia_Pre_Insuficiencia', 'Qtd_Procedimentos_Pre_Insuficiencia']
    df_kpi_preventivo_insuficiencia_COMBINED[cols_kpi] = df_kpi_preventivo_insuficiencia_COMBINED[cols_kpi].fillna(0).astype(int)

    print("\n--- 🚨 Métricas Preventivas por Paciente (Comportamento Pré-Insuficiência) ---")
    display(df_kpi_preventivo_insuficiencia_COMBINED.head())

print("✅ Célula 9 concluída. Tabela 'df_kpi_preventivo_insuficiencia_COMBINED' criada.")

print("\n==================================================================================")
print("🏁 PIPELINE (COMBINADO 4/5) PARA INSUFICIÊNCIA CONCLUÍDO 🏁")
print("==================================================================================")
```

```python
# =============================================================================
# Célula 10 (FINAL CORRIGIDA): Tabela Comparativa (Período Completo)
# =============================================================================
import pandas as pd
import numpy as np

print("🚀 Iniciando Célula 10 (CORRIGIDA): Consolidação dos KPIs de todas as 7 Doenças (Período Completo)...")

# 1. Lista de todos os DataFrames de KPIs preventivos criados
kpi_dataframes = [
    ('Infarto', 'df_kpi_preventivo_infarto_COMBINED', 'Infarto'),
    ('Angina', 'df_kpi_preventivo_angina_COMBINED', 'Angina'),
    ('Arritmia', 'df_kpi_preventivo_arritmia_COMBINED', 'Arritmia'),
    ('Insuficiência', 'df_kpi_preventivo_insuficiencia_COMBINED', 'Insuficiencia'),
    ('Aneurisma', 'df_kpi_preventivo_aneurisma_COMBINED', 'Aneurisma'),
    ('AVC', 'df_kpi_preventivo_avc_COMBINED', 'AVC'),
    ('Hipertensão', 'df_kpi_preventivo_hipertensao_COMBINED', 'Hipertensao')
]

# 2. Criar uma única tabela com todos os KPIs
df_kpi_consolidado_list = []

for condicao, df_nome, sufixo_doenca in kpi_dataframes:
    try:
        df_kpi = globals()[df_nome].copy()

        # 2.1 Mapear colunas originais para nomes genéricos
        mapping = {
            f'Qtd_Cardio_Eletiva_Pre_{sufixo_doenca}': 'Qtd_Cardio_Eletiva_Pre',
            f'Qtd_Emergencia_Pre_{sufixo_doenca}': 'Qtd_Emergencia_Pre',
            f'Qtd_Procedimentos_Pre_{sufixo_doenca}': 'Qtd_Procedimentos_Pre',
        }

        df_kpi_temp = df_kpi.rename(columns=mapping)
        df_kpi_temp['Condição'] = condicao
        df_kpi_consolidado_list.append(df_kpi_temp)

    except KeyError:
        print(f"  - AVISO: DataFrame '{df_nome}' para '{condicao}' não foi encontrado ou colunas de KPI estão erradas. A pular.")

if not df_kpi_consolidado_list:
    print("\nERRO: Nenhuma tabela de KPIs foi encontrada. Não é possível gerar o resumo.")
else:
    df_kpi_consolidado = pd.concat(df_kpi_consolidado_list, ignore_index=True)

    # 3. Agrupar pela Condição e GRUPO_ANGIO e calcular as médias finais
    print("   - Calculando médias finais por Condição e Grupo Angio...")

    # Colunas de KPI genéricas após renomeação
    kpi_cols_genericas = ['Qtd_Cardio_Eletiva_Pre', 'Qtd_Emergencia_Pre', 'Qtd_Procedimentos_Pre']

    df_comparativo_final_COMBINED = df_kpi_consolidado.groupby(['Condição', 'GRUPO_ANGIO']).agg(
        N_Pacientes=('ID_PESSOA', 'size'),
        Média_Consultas_Cardio_Eletivas=('Qtd_Cardio_Eletiva_Pre', 'mean'),
        Média_Visitas_à_Emergência=('Qtd_Emergencia_Pre', 'mean'),
        Média_Procedimentos_Chave=('Qtd_Procedimentos_Pre', 'mean')
    ).reset_index()

    # 4. Formatar e exibir
    df_comparativo_final_COMBINED = df_comparativo_final_COMBINED.rename(columns={'N_Pacientes': 'Nº Pacientes'})
    df_comparativo_final_COMBINED = df_comparativo_final_COMBINED.sort_values(by=['Condição', 'GRUPO_ANGIO'])

    # --- Exibir Tabela Final ---
    print("\n\n" + "="*80)
    print("--- 🚀 TABELA COMPARATIVA FINAL (Período Completo) POR GRUPO DE ANGIOTOMOGRAFIA ---")
    print("="*80)
    display(df_comparativo_final_COMBINED.round(2))
    print("✅ Célula 10 concluída.")
```

```python
# =============================================================================
# Célula 11 (FINAL CORRIGIDA): Tabela Comparativa (Janelas de Tempo)
# (Para o pipeline combinado)
# =============================================================================
import pandas as pd
import numpy as np

print("🚀 Iniciando Célula 11: Análise Comparativa por JANELAS DE TEMPO (Combinado)...")

# 1. Definição dos DataFrames para loop
analises_a_fazer = [
    ('Infarto', 'df_timeline_final_view_infarto_COMBINED'),
    ('Angina', 'df_timeline_final_view_angina_COMBINED'),
    ('Arritmia', 'df_timeline_final_view_arritmia_COMBINED'),
    ('Insuficiência', 'df_timeline_final_view_insuficiencia_COMBINED'),
    ('Aneurisma', 'df_timeline_final_view_aneurisma_COMBINED'),
    ('AVC', 'df_timeline_final_view_avc_COMBINED'),
    ('Hipertensão', 'df_timeline_final_view_hipertensao_COMBINED')
]

# 2. Lista de Janelas de Tempo
time_windows = [
    ('Último 1 Ano', 365),
    ('Últimos 2 Anos', 730),
    ('Período Completo', None)
]

# --- 3. Criar a função auxiliar de cálculo de janelas ---
def calcular_janelas_vectorizado_final(df_timeline, nome_doenca):
    resultados_janela = []

    for window_name, dias in time_windows:

        # 1. Filtrar eventos pré-evento DENTRO da janela de tempo
        if dias is None:
            df_pre = df_timeline[df_timeline['Dias_Desde_Evento_0'] < 0].copy()
        else:
            df_pre = df_timeline[
                (df_timeline['Dias_Desde_Evento_0'] < 0) &
                (df_timeline['Dias_Desde_Evento_0'] >= -dias)
            ].copy()

        # 2. Base de pacientes (ID_PESSOA e GRUPO_ANGIO)
        df_pacientes = df_timeline[['ID_PESSOA', 'GRUPO_ANGIO']].drop_duplicates()

        # 3. Calcular KPIs (agrupando por ID e GRUPO_ANGIO)
        kpi_cardio_pre = df_pre[df_pre['Evento_Label'] == 'VISITA: Cardio Eletiva'].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Cardio_Eletiva')
        kpi_emerg_pre = df_pre[df_pre['Evento_Label'] == 'VISITA: Emergência'].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Emergencia')
        kpi_proc_pre = df_pre[df_pre['Evento_Label'].str.contains('PROC:')].groupby(['ID_PESSOA', 'GRUPO_ANGIO']).size().reset_index(name='Qtd_Procedimentos')

        # 4. Consolidar no nível do Paciente
        df_kpi_paciente = df_pacientes.merge(kpi_cardio_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')
        df_kpi_paciente = df_kpi_paciente.merge(kpi_emerg_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')
        df_kpi_paciente = df_kpi_paciente.merge(kpi_proc_pre, on=['ID_PESSOA', 'GRUPO_ANGIO'], how='left')

        cols_kpi = ['Qtd_Cardio_Eletiva', 'Qtd_Emergencia', 'Qtd_Procedimentos']
        df_kpi_paciente[cols_kpi] = df_kpi_paciente[cols_kpi].fillna(0).astype(int)

        # 5. Calcular Média Final por GRUPO_ANGIO
        df_medias = df_kpi_paciente.groupby('GRUPO_ANGIO').agg(
            N_Pacientes=('ID_PESSOA', 'size'),
            Média_Consultas_Cardio_Eletivas=('Qtd_Cardio_Eletiva', 'mean'),
            Média_Visitas_à_Emergência=('Qtd_Emergencia', 'mean'),
            Média_Procedimentos_Chave=('Qtd_Procedimentos', 'mean')
        ).reset_index()

        df_medias['Condição'] = nome_doenca
        df_medias['Janela de Tempo'] = window_name

        resultados_janela.append(df_medias)

    return pd.concat(resultados_janela, ignore_index=True)


# --- 4. Executar o Loop Principal ---

resultados_janelas_list = []

for nome, df_nome in analises_a_fazer:
    try:
        df_real = globals()[df_nome]
        print(f"   - Calculando janelas para {nome}...")
        df_janelas_doenca = calcular_janelas_vectorizado_final(df_real, nome)
        resultados_janelas_list.append(df_janelas_doenca)
    except KeyError:
        print(f"   - AVISO: DataFrame '{df_nome}' não foi encontrado. A pular {nome}.")

# 5. Concatenar o resultado final e formatar
if not resultados_janelas_list:
    print("\nERRO: Nenhuma tabela de janelas foi criada. Não é possível gerar o resumo.")
else:
    df_comparativo_final_janelas_COMBINED = pd.concat(resultados_janelas_list, ignore_index=True)

    df_comparativo_final_janelas_COMBINED = df_comparativo_final_janelas_COMBINED.rename(columns={'N_Pacientes': 'Nº Pacientes'})
    df_comparativo_final_janelas_COMBINED = df_comparativo_final_janelas_COMBINED.set_index(['Condição', 'GRUPO_ANGIO', 'Janela de Tempo'])

    # --- Exibir Tabela Final ---
    print("\n\n" + "="*80)
    print("--- 🚀 MATRIZ FINAL DE RISCO: KPI PREVENTIVO por GRUPO DE ANGIOTOMOGRAFIA ---")
    print("--- (Compara 7 Doenças x 5 Grupos Angio x 3 Janelas de Tempo) ---")
    print("="*80)

    df_comparativo_final_janelas_COMBINED = df_comparativo_final_janelas_COMBINED.sort_index(level=[0, 1, 2])

    pd.set_option('display.max_rows', 500)
    pd.set_option('display.max_columns', 500)
    display(df_comparativo_final_janelas_COMBINED.round(2))
    pd.reset_option('display.max_rows')
    pd.reset_option('display.max_columns')

print("\n✅ Análise comparativa (Célula 11) concluída. Este é o resultado final do teu pipeline!")
```
